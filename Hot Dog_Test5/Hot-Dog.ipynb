```python
import os
import pandas as pd
from teachable_machine import TeachableMachine
from IPython.display import display

# 1. Загружаем модель (убедитесь, что имя файла совпадает)
model = TeachableMachine(model_path="keras_Model.h5")

# 2. Функция предсказания
def predict_image(image_path):
    result, _ = model.classify_and_show(image_path)
    # Возвращаем очищенное от цифры название и уверенность
    top_class = result["class_name"]
    clean_class = top_class[2:] if top_class[0].isdigit() else top_class
    return clean_class.strip(), result["class_confidence"]

# 3. Функция проверки папки
def evaluate_folder(folder_path, expected_label):
    results = []
    if not os.path.exists(folder_path):
        print(f"⚠️ Папка '{folder_path}' не найдена. Проверьте путь.")
        return pd.DataFrame()

    # Считаем только картинки
    images = [f for f in sorted(os.listdir(folder_path)) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    
    for filename in images:
        file_path = os.path.join(folder_path, filename)
        pred_label, conf = predict_image(file_path)
        
        # Сравниваем без учета регистра
        is_correct = "✅ Верно" if pred_label.lower() == expected_label.lower() else "❌ Ошибка"
        
        results.append({
            "Файл": filename,
            "Ожидаемый": expected_label,
            "Предсказано": pred_label,
            "Уверенность": f"{conf:.2%}",
            "Результат": is_correct
        })
    return pd.DataFrame(results)
```

```python
# 4. Запускаем проверку
print("--- Проверка: Hot-dogs ---")
# Обратите внимание: ожидаемый класс пишем так, как он записан в labels.txt (английскими буквами)
df_hot = evaluate_folder("hot_dog_validation", "Hot-dogs")
display(df_hot)
```

```python
print("\n--- Проверка: Not Hot-dogs ---")
df_not_hot = evaluate_folder("not_hot_dog_validation", "Not Hot-dogs")
display(df_not_hot)
```

```python

# 5. Вывод итоговой точности
if not df_hot.empty:
    acc_hot = (df_hot["Результат"] == "✅ Верно").mean() * 100
    print(f"\nИтоговая точность на хот-догах: {acc_hot:.1f}%")

if not df_not_hot.empty:
    acc_not_hot = (df_not_hot["Результат"] == "✅ Верно").mean() * 100
    print(f"Итоговая точность на не хот-догах: {acc_not_hot:.1f}%")
```
