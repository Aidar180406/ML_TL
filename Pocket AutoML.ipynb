<h2>Ограничения On-Device ML (на примере Pocket AutoML):</h2>

При добавлении новой порции фотографий (более 10 штук за раз) приложение Pocket AutoML на смартфоне Samsung Galaxy A31 начинало заметно зависать.

Причина: В отличие от Teachable Machine, где обучение происходит на мощных серверах Google в облаке, Pocket AutoML выполняет все вычисления локально на устройстве. Бюджетный процессор Galaxy A31 не справляется с быстрой обработкой десятков новых изображений и пересчетом весов модели. Это приводит к переполнению оперативной памяти и временному зависанию интерфейса.

Вывод: Для комфортной работы с Pocket AutoML на слабых устройствах рекомендуется добавлять фотографии небольшими партиями (по 3–5 штук), давая приложению время на обработку, либо использовать более производительный смартфон. Несмотря на задержки, приложение успешно выполняет свою задачу и демонстрирует возможности TensorFlow Lite для работы без интернета.

<h2>Результаты дообучения модели в Pocket AutoML:</h2>

В процессе работы с приложением я добавлял новые изображения в классы, исправляя ошибки распознавания, но зависает приложение

Приложение автоматически отслеживает изменения. После добавления новых данных оно выдало рекомендацию «перетренировать модель». После повторного обучения (вкладка «Модель») точность распознавания сохранилась на высоком уровне — 95% достоверности.

Это демонстрирует главное преимущество Pocket AutoML: возможность непрерывного обучения (Continuous Learning). Пользователь может улучшать модель прямо во время её использования, просто подтверждая или исправляя результаты, без необходимости заново собирать весь датасет.
