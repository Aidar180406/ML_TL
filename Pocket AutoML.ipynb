<h2>Ограничения On-Device ML (на примере Pocket AutoML):</h2>

При добавлении новой порции фотографий (более 10 штук за раз) приложение Pocket AutoML на смартфоне Samsung Galaxy A31 начинало заметно зависать.

Причина: В отличие от Teachable Machine, где обучение происходит на мощных серверах Google в облаке, Pocket AutoML выполняет все вычисления локально на устройстве. Бюджетный процессор Galaxy A31 не справляется с быстрой обработкой десятков новых изображений и пересчетом весов модели. Это приводит к переполнению оперативной памяти и временному зависанию интерфейса.

Вывод: Для комфортной работы с Pocket AutoML на слабых устройствах рекомендуется добавлять фотографии небольшими партиями (по 3–5 штук), давая приложению время на обработку, либо использовать более производительный смартфон. Несмотря на задержки, приложение успешно выполняет свою задачу и демонстрирует возможности TensorFlow Lite для работы без интернета.

<h2>Результаты дообучения модели в Pocket AutoML:</h2>

В процессе работы с приложением я добавлял новые изображения в классы, исправляя ошибки распознавания, но зависает приложение

Приложение автоматически отслеживает изменения. После добавления новых данных оно выдало рекомендацию «перетренировать модель». После повторного обучения (вкладка «Модель») точность распознавания сохранилась на высоком уровне — 95% достоверности.

Это демонстрирует главное преимущество Pocket AutoML: возможность непрерывного обучения (Continuous Learning). Пользователь может улучшать модель прямо во время её использования, просто подтверждая или исправляя результаты, без необходимости заново собирать весь датасет.

### Мини-эксперимент: бинарная классификация (2 класса)

В дополнение к основному эксперименту был проведен мини-эксперимент с использованием двух классов («Свои» и «Чужие») и сокращенного набора данных (21 тренировочное изображение). Целью данного этапа являлась проверка работоспособности пайплайна Pocket AutoML при минимальных требованиях к объему обучающей выборки.

#### Метрики точности:
Согласно данным, отображенным на вкладке «Модель», точность классификации составила диапазон от 70.84% до 97.09% при 95% достоверности. Несмотря на малый объем тренировочной выборки, модель продемонстрировала способность к обобщению: на этапе тестирования (вкладка «Распознать») автомобиль был корректно классифицирован как «Свой».

#### Анализ результатов:
Полученные результаты подтверждают, что даже при ограниченном наборе данных (порядка 10 изображений на класс) приложение Pocket AutoML способно обучить работоспособную модель. Это демонстрирует низкий порог входа для создания пользовательских ML-моделей на мобильных устройствах и подтверждает эффективность технологии TensorFlow Lite для решения задач классификации в условиях ограниченных вычислительных ресурсов.

### Общий вывод по эксперименту с Pocket AutoML:

* В ходе выполнения задания было освоено мобильное приложение Pocket AutoML, предназначенное для обучения моделей классификации изображений непосредственно на устройстве пользователя (On-Device ML). Эксперимент проводился на смартфоне Samsung Galaxy A31.

* В рамках основного эксперимента было создано три класса объектов, проведено семь циклов тестирования с подтверждением и исправлением результатов. Выявлено ограничение производительности: при добавлении более 10 изображений за один сеанс приложение испытывало перегрузку, что проявлялось в зависании интерфейса. Данный эффект объясняется ограниченными вычислительными ресурсами бюджетного смартфона.

* Несмотря на выявленные ограничения, после дообучения (увеличение обучающей выборки со 179 до 196 изображений) точность модели сохранилась на уровне 95%, что подтверждает эффективность механизма непрерывного обучения (Continuous Learning). Дополнительный мини-эксперимент (2 класса, 10 изображений) продемонстрировал работоспособность приложения даже при минимальном объеме данных.

* Таким образом, Pocket AutoML представляет собой эффективный инструмент для быстрого создания прототипов ML-моделей без необходимости использования облачной инфраструктуры и без передачи данных за пределы устройства, что обеспечивает конфиденциальность. Однако для комфортной работы на слабых устройствах рекомендуется оптимизировать процесс загрузки данных (небольшие партии) или использовать более производительное оборудование.
