```python
import os
import pandas as pd
from teachable_machine import TeachableMachine
from IPython.display import display


# 2. Загружаем модель и явно указываем файл меток
# ВАЖНО: проверьте, как точно называется ваш файл модели: keras_Model.h5 или keras_model.h5
model = TeachableMachine(
    model_path="keras_Model.h5",  # <-- Исправьте на маленькую букву, если у вас keras_model.h5
    labels_file_path="labels.txt" # <-- Явно передаем путь к меткам
)

# 3. Функция предсказания
def predict_image(image_path):
    result, _ = model.classify_and_show(image_path)
    top_class = result["class_name"]
    # Очищаем от цифры (например, "0 Иностранные" -> "Иностранные")
    clean_class = top_class[2:] if top_class[0].isdigit() else top_class
    return clean_class.strip(), result["class_confidence"]

# 4. Проверка папки
folder_path = r"D:\Автомобили\Тестирование" 
results = []

if not os.path.exists(folder_path):
    print(f"⚠️ Папка '{folder_path}' не найдена.")
else:
    images = [f for f in sorted(os.listdir(folder_path)) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    print(f"Найдено изображений: {len(images)}. Начинаем проверку...\n")

    for filename in images:
        file_path = os.path.join(folder_path, filename)
        pred_label, conf = predict_image(file_path)
        
        results.append({
            "Файл": filename,
            "Предсказано моделью": pred_label,
            "Уверенность": f"{conf:.2%}"
        })

    df_results = pd.DataFrame(results)
    display(df_results)
```

```python
if not df_results.empty:
    acc_hot = (df_results["Предсказано моделью"] == "russian").mean() * 100
    print(f"\nИтоговая точность на Российские: {acc_hot:.1f}%")

if not df_results.empty:
    acc_hot = (df_results["Предсказано моделью"] == "foreign").mean() * 100
    print(f"\nИтоговая точность на Иностранные: {acc_hot:.1f}%")

if not df_results.empty:
    acc_hot = (df_results["Предсказано моделью"] == "trucks").mean() * 100
    print(f"\nИтоговая точность на Грузовики: {acc_hot:.1f}%")
```

<h3>Общий вывод:</h3> Модель показала высокую эффективность на тестовой выборке.

- Класс «Грузовики» распознается практически безошибочно (уверенность 78–96%). Модель без труда определяет Isuzu, MAN, Volvo и Scania благодаря их характерным крупным размерам и форме.

- Класс «Иностранные» также показывает хорошие результаты. Модель уверенно определяет BMW и Bentley (уверенность 60–82%).

- Класс «Российские» оказался самым сложным. Модель иногда путала старые отечественные автомобили (например, ВАЗ-2107 на фото vaz-2107-lada-6.jpg) с иностранными. Это связано с тем, что старые фото часто имеют низкое разрешение, а классический дизайн «Жигулей» может перекликаться с европейскими ретро-автомобилями.
