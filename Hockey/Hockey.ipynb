```python
import os
import pandas as pd
from teachable_machine import TeachableMachine
from IPython.display import display

model = TeachableMachine(
    model_path="keras_model.h5",  
    labels_file_path="labels.txt" 
)

# 3. Функция предсказания
def predict_image(image_path):
    result, _ = model.classify_and_show(image_path)
    top_class = result["class_name"]
    clean_class = top_class[2:] if top_class[0].isdigit() else top_class
    return clean_class.strip(), result["class_confidence"]

# 4. Проверка папки
folder_path = r"D:\Хоккеисты\Тестированиеэ" 
results = []

if not os.path.exists(folder_path):
    print(f"⚠️ Папка '{folder_path}' не найдена.")
else:
    images = [f for f in sorted(os.listdir(folder_path)) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
    print(f"Найдено изображений: {len(images)}. Начинаем проверку...\n")

    for filename in images:
        file_path = os.path.join(folder_path, filename)
        pred_label, conf = predict_image(file_path)
        
        results.append({
            "Файл": filename,
            "Предсказано моделью": pred_label,
            "Уверенность": f"{conf:.2%}"
        })

    df_results = pd.DataFrame(results)
    display(df_results)
```

```python
if not df_results.empty:
    acc_hot = (df_results["Предсказано моделью"] == "Salavat Yulaev").mean() * 100
    print(f"\nИтоговая точность на Салавата Юлаева: {acc_hot:.1f}%")

if not df_results.empty:
    acc_hot = (df_results["Предсказано моделью"] == "Judges").mean() * 100
    print(f"\nИтоговая точность на Судьи: {acc_hot:.1f}%")

if not df_results.empty:
    acc_hot = (df_results["Предсказано моделью"] == "Guests").mean() * 100
    print(f"\nИтоговая точность на Гости: {acc_hot:.1f}%")
```

<h3>Общий вывод:</h3> Эксперимент успешно удался! 

- Однако модель часто ошибается на изображениях с общими планами, где на льду одновременно находятся игроки разных команд, судьи и видна толпа. В таких случаях модель теряет фокус и выдает низкую уверенность (40–50%), путая классы.

- Также были замечены ошибки на фотографиях, где хоккеисты одеты в форму других цветов (например, синюю), но имеют крупные номера на спине — модель ошибочно относила их к «Салавату Юлаеву».

- Причина: Обучающая выборка, вероятно, состояла в основном из крупных планов игроков в зеленой форме. Модель не научилась игнорировать фон и других людей, поэтому на сложных сценах она теряется.
