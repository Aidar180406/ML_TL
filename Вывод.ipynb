<h1>Эксперимент «Pocket AutoML» (Задание 3)</h1>
Ход работы:

Установлено приложение Pocket AutoML на смартфон Samsung Galaxy A31.

- Созданы 3 класса объектов и добавлено минимум по 10 изображений в каждый.

- Проведено 7 тестов с подтверждением/исправлением результатов.

- Выполнено дообучение модели: количество тренировочных изображений увеличено со 179 до 196. После повторной тренировки точность сохранилась на уровне 95% достоверности.

Наблюдения за производительностью:

Смартфон Galaxy A31 — бюджетное устройство. При добавлении новой порции фотографий (более 10 штук за раз) приложение начинало зависать.

- Причина: Pocket AutoML выполняет все вычисления локально на устройстве (On-Device ML), в отличие от Teachable Machine, где обучение идет в облаке. Бюджетный процессор не справляется с быстрой обработкой десятков изображений и пересчетом весов модели.

Вывод: Приложение успешно решает задачу обучения моделей прямо на телефоне. Для комфортной работы на слабых устройствах рекомендуется добавлять фотографии небольшими партиями (по 3–5 штук). Главное преимущество Pocket AutoML — возможность непрерывного обучения (Continuous Learning).

<h1>Эксперимент «FastAPI + Streamlit» (Задание 4)</h1>
Ход работы:

- Разработан API-сервер на FastAPI, который загружает модель .h5 при старте и принимает POST-запросы с изображениями.

- Разработан интерфейс на Streamlit, позволяющий пользователю загрузить картинку и получить результат распознавания.

- В ходе разработки решены технические проблемы:

- Ошибка numpy.float32 при сериализации JSON — исправлена приведением к float.

- Проблема с кодировкой labels.txt (русские буквы) — решена пересохранением файла в UTF-8.

- Ошибка DepthwiseConv2D при загрузке .h5 — решена патчем совместимости Keras.

Вывод: Создана полноценная клиент-серверная архитектура. Разделение интерфейса (Streamlit) и модели (FastAPI) позволяет легко масштабировать систему.

<h1>Финальный вывод по всей работе</h1>
Общий вывод по результатам выполнения всех заданий

В ходе работы были освоены три ключевых инструмента современного Machine Learning: облачный сервис Teachable Machine, мобильное приложение Pocket AutoML и фреймворки для развертывания моделей FastAPI + Streamlit.

1. Качество моделей (Teachable Machine):
Наилучший результат показала модель «Хот-доги» с общей точностью 91.1%. Модели «Хоккей» и «Автомобили» показали более скромные результаты (средняя уверенность 60–80%). Главная причина — сложность и неоднородность тестовой выборки. Модели отлично справляются с крупными планами объектов (уверенность 90%+), но теряются на общих планах, размытых фото и объектах с необычным ракурсом.

2. Ограничения On-Device ML (Pocket AutoML):
При работе с Pocket AutoML на смартфоне Galaxy A31 выявлена ключевая проблема — нехватка вычислительных ресурсов. При добавлении более 10 фотографий за раз приложение зависало. Это наглядно демонстрирует разницу между облачными вычислениями (Teachable Machine) и локальным обучением (On-Device ML). Тем не менее, приложение успешно решает задачу и поддерживает непрерывное обучение (Continuous Learning).

3. Техническая реализация (FastAPI + Streamlit):
Разработана полноценная клиент-серверная архитектура. API на FastAPI загружает модель один раз при старте и обрабатывает запросы, а интерфейс на Streamlit позволяет пользователю загружать изображения. В процессе разработки решены реальные технические проблемы: сериализация numpy.float32, кодировка labels.txt, совместимость версий Keras (DepthwiseConv2D). Это подтверждает, что модель можно легко интегрировать в любое веб-приложение.

4. Главный урок:
Качество модели машинного обучения напрямую зависит от разнообразия обучающей выборки. Если в обучающем наборе преобладают крупные планы объектов, модель будет ошибаться на общих сценах. Для повышения точности необходимо добавлять больше разнообразных данных: разные ракурсы, освещение, фоны и масштабы. Это подтверждают ошибки в хоккее (общие планы) и автомобилях (старые фото ВАЗ).
